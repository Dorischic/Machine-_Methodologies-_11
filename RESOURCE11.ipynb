```python
# Ask a question about the data
# Try to answer a question you have about the data using a plot or statistics
# what are the locations with the fastest roller coasters with a minimum of 10?
```

```python
df['Location'].value_counts()
```

```python
df.query('Location != "other"')

```

```python
df.query('Location != "other"') \
  .groupby('Location')['speed_mph'] \
  .agg(['mean','count'])
```

```python
df.query('Location != "other"') \
  .groupby('Location')['speed_mph'] \
  .agg(['mean','count']) \
  .query('count >= 10')
```

```python
df.query('Location != "other"') \
  .groupby('Location')['speed_mph'] \
  .agg(['mean','count']) \
  .query('count >= 10') \
  .sort_values('mean')
```

```python
ax= df.query('Location != "other"') \
  .groupby('Location')['speed_mph'] \
  .agg(['mean','count']) \
  .query('count >= 10') \
  .sort_values('mean')['mean'] \
  .plot(kind='barh', figsize=(12,5), title='Average coaster speed by Location')

ax.set_xlabel('Average coaster speed')
plt.show()
```
